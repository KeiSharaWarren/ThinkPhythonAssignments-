```python
# ==== STRUCTURE & PSEUDOCODE ====

# write_entry function
# - open file in append mode
# - add the journal entry to the file
# - make sure it goes on a new line
# - close the file

# read_journal function
# - open file in read mode
# - read everything in the file
# - display it to the user
# - if file doesn't exist, print a message

# main program
# - Set filename to journal.txt
# - use a loop so program keeps running
# - ask user if they want to write, read, or quit
# - if write:
#     - get entry from user
#     - call write_entry
# - if read:
#     - call read_journal
# - if quit:
#     - stop program
# - otherwise:
#     - print invalid option

# ==== MIDPOINT CODE ====

def write_entry(filename, entry):
    with open(filename, 'a') as file:
        file.write(entry + '\n')

def read_journal(filename):
    with open(filename, 'r') as file:
        contents = file.read()
        print(contents)

filename = "journal.txt"

while True:
    choice = input("Enter W to write, R to read, or Q to quit: ").lower()

    if choice == 'w':
        entry = input("Enter your journal entry:")
        write_entry(filename, entry)

    elif choice == 'r':
        read_journal(filename)

    elif choice == 'q':
        break

    else: print("Invalid option ")

# ==== FINAL CODE ====

def write_entry(filename, entry):
    with open(filename, 'a') as file:
        file.write(entry + '\n')

def read_journal(filename):
   try:
        with open(filename, 'r') as file:
            contents = file.read()
            print(contents)
   except  FileNotFoundError:
       print("No journal entries found.")

filename = "journal.txt"

while True:
    choice = input("Enter W to write, R to read, or Q to quit:").lower()

    if choice == 'w':
        entry = input("Enter your journal entry:")
        write_entry(filename, entry)

    elif choice == 'r':
        read_journal(filename)

    elif choice == 'q':
        break

    else:
        print("Invalid option")
```
