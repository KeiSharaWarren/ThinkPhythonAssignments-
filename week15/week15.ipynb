```python
AI_BASE_Build

Goal:
- The goal is to create a personal task and priority manager where users can keep track of their tasks.


Plan:
- Use a list to store tasks
- Each task will have a name and a priority level (High, Medium, or Low)

Features:
1. Add Task
- Ask the user to enter a task
- Ask for the priority level
_ Save the task in the list

2. View Tasks
   -Show all tasks along with their priority

3. Complete Task
   - Let the user choose a task
   - Mark it as completed or remove

4. Exit
  - End the program when the user chooses to exit

  Logic Flow:
  - Start the program
  - Display a menu of options
  - Carry out the option the user selects
  - Repeat until the user chooses to exit




```

```python
tasks = []

while True:
    print("\n1. Add Task")
    print("2.  View Tasks")
    print("3.  Complete Task")
    print("4.  What's next?")
    print("5. Exit")

    choice = input("Choose an option:")

    if choice == "1":
        task = input("Enter a task:")
        tasks.append(task)
        print ("Task added.")

    elif choice == "2":
        if len(tasks) == 0:
            print("No tasks yet. ")
        else:
            for i in range(len(tasks)):
                print(i, "-", tasks[i])

    elif choice == "3":
        if len(tasks) == 0:
            print("No tasks to complete.")
        else:
            for i in range(len(tasks)):
                print(i, "-", tasks[i])

            index = int(input("Enter the number of the task to complete"))
            completed_task = tasks.pop(index)

            with open("history.txt", "a") as file:
                file.write(completed_task + "\n")

            print("Tasks completed and saved.")


    elif choice == "4":
        if len(tasks)==0:
            print("No tasks avalible.")
        else:
            print("Next task to work on:", tasks[0])

    elif choice == "5": # Added colon here
        print("Goodbye!")
        break

    else: # This else handles invalid choices, moved to the end
        print("Invalid option. Please try again.")
```

LOGIC_REFINEMENT

I added a feature called "what's next?" that shows the next tasks in  the list.
If there are no tasks, the program displays a message

I also fixed the program logic so that all menu options work correctly,
including exiting the program when the user selects option 5.
