```python
tasks = []

while True:
    print("\n1. Add Study Session")
    print("2. View Study Sessions")
    print("3. Exit")

    choice = input("Choose an option:")

    if choice == "1":
        subject = input("Enter subject: ")
        tasks.append(subject)
        print("Session added.")

    elif choice == "2":
        if len(tasks) == 0:
            print("No sessions yet.")
        else:
            for i in range(len(tasks)):
                print(i, "-", tasks[i])

    elif choice == "3":
        print("Goodbye!")
        break

    else:
        print("Invalid option")
```
