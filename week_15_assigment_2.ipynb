```python
tasks = []

try:
    with open("sessions.txt", "r") as file:
        for line in file:
            tasks.append(line.srip())
except:
    pass

while True:
    print("1. Add Study Session")
    print("2. View Study Sessions")
    print("3. Exit")

    choice = input("Choose an option:")

    if choice == "1":
        task = input("Enter a study session: ")
        tasks.append(task)

        with open("sessions.txt", "a") as file:
            file.write(task + "\n")

        print("session added.")

    elif choice == "2":
        if len(tasks) ==0:
            print("No sessions yet.")
        else:
            for i in range(len(tasks)):
                print(i, "-", tasks[i])

    elif choice == "3":
        print("Goodbye!")
        break

    else:
        print("Invalid option")





```
