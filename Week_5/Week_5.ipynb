```python
#IMPORT STATEMENTS
#(None)

#FUNCTION DEFINITIONS
def calculate_ticket(age, hour):
    if age <=12 or age >= 65:
       price = 8
    else:
        price = 12
    if hour < 17:
        price -= 2

    return price

#PROGRAM LOGIC
#1. input() for age (int)
#2. input() for movie hour (int)
#3. call calculate_ticket()

age_input = int(input("Enter age: "))
hour_input = int(input("Enter movie hour (24hr format):"))

final_price = calculate_ticket(age_input, hour_input)
print("Final ticket price:", final_price)
```
