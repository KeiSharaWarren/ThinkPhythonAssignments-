```python
expenses = []

while True:
    print("\n--- Expense Tracker ---")
    print("1. Add New Expense")
    print("2. Get Total Expenses")
    print("3. Exit")

    choice = input("Selection:")

    if choice == "1":
        category = input("Enter Category: ")
        vendor = input("Enter Vendor: ")

        try:
            amount = float(input("Enter Amount "))
        except ValueError:
            print("Invalid amount")
            continue

        date = input("Enter Date: ")

        expense = {
            "category": category,
            "vendor": vendor,
            "amount": amount,
            "date": date
        }

        expenses.append(expense)

        print("Expense added succesfully.")

    elif choice == "2":

        print("\n--- Totaling Options ---")
        print("A: All Expenses")
        print("C: Filter by Category")
        print("V: Filter  by Vendor")

        total_choice = input("Choice:").upper()

        if total_choice == "A":
            total= 0

            for expense in expenses:
                total += expense["amount"]

            print(f"Total Expenses: ${total:.2f}")

        elif total_choice == "C":
            search_category = input("Enter Category to sum")

            total= 0

            for expense in expenses:
                if expense["category"].lower() == search_category.lower():
                   total += expense["amount"]

            print(f"Total Expenses: ${total:.2f}")

        elif total_choice == "V":
            search_vendor = input("Enter Vendor to sum:")

            total = 0

            for expense in expenses:
                if expense["category"].lower() == search_category.lower():
                    total += expense["amount"]

            print(f"The total for {search_vendor} is: ${total:.2f}")

        else:
           print("Invalid option")

    elif choice == "3":
         print("Goodbye.")
         break

    else:
         print("Invalid selection")





```
