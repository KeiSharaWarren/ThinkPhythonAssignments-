```python
# NAME: gradebook.py
# IMPORT STATEMENTS

# FUNCTON DDEFINITIONS
def add_students(gradebook):
    while True:
        name = input("Enter student's name (or 'done'): ")

        if name == 'done':
            break

        grade = int(input(f"Enter {name}'s grade: "))
        gradebook[name] = grade

# PROGRAM LOGIC
gradebook = {}

add_students(gradebook)

print(gradebook)
```
