```python
# NAME: gradebook.py
# IMPORT STATEMENTS

# FUNCTON DDEFINITIONS
def add_students(gradebook):
    while True:
        name = input("Enter student's name (or 'done'): ")

        if name == 'done':
            break

        grade = int(input(f"Enter {name}'s grade: "))
        gradebook[name] = grade

def calculate_average(gradebook):
    if len(gradebook) ==0:
        return 0
    total = sum(gradebook.values())
    average = total / len(gradebook)
    return average

# PROGRAM LOGIC
gradebook = {}

add_students(gradebook)

print("--- Class Report ---")
for name, grade in gradebook.items():
    print(f"{name}: {grade}")

average = calculate_average(gradebook)
print(f"Class Average: {average:.2f}")
```
