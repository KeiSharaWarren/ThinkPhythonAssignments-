```python
class Book:
  def __init__(self, title, author):
      # save book info
      self.title = title
      self.author = author
      self.is_available = True

  def __str__(self):
      # display book info
      if self.is_available:
          status = "Available"
      else:
          status = "Checked Out"
      return self.title + " by " + self.author + "-" + status

  def checkout(self):
      # mark as checked out
      self.is_available = False

  def return_book(self):
      # mark as returned
      self.is_available = True

# PROGRAM LOGIC

book_one = Book("Think phython", "Allen B. Downey")
book_two = Book("The Hobbit", "J.R.R. Tolkien")

print("Library Status:")
print(book_one)
print(book_two)

print("\nChecking out The Hobbit...\n")
book_two.checkout()

print("Updated Library Status:")
print(book_one)
print(book_two)




```
