```python
# Import Statements
#(None)

# Function Definitions
# Define print_bio with three parameters: name, age and goal
def print_bio(name, age, goal):
    # Inside the function use string multipication (*) for a border line
    border = "*" * 30
    print(border)
    # Print the bio details using the parameter names
    print(f"Name: {name}")
    print(f"Age: {age}")
    print(f"Goal: {goal}")
    print(border)

# Define variables for personal info (using existing kernel variables)
my_name = name
my_age = age
my_goal = goal

# Define variables for famous person
famous_person_name = "PinkPatheress"
famous_person_age = 24
famous_person_goal = "Make great music"

# Define variables for fictional character
fictional_character_name = "Twilight Sparkle"
fictional_character_age = 22
fictional_character_goal = "study magic of friendship"

# Program Logic
# Call print_bio using my own info
print_bio(my_name, my_age, my_goal)

# Call print_bio using (Name of Famous Person)
print_bio(famous_person_name, famous_person_age, famous_person_goal)

# Call print_bio using (Name of fictional character)
print_bio(fictional_character_name, fictional_character_age, fictional_character_goal)
```
