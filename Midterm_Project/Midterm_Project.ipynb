```python
# NAME: rentalcar.py

# FUNCTION DEFINITONS

def get_classification_name(code):
    if code == "B":
        return "Budget"
    elif code == "S":
        return "Standard"
    elif code == "L":
        return "Luxury"

def get_daily_rate(code):
    if code == "B":
        return 40
    elif code == "S":
        return 60
    elif code == "L":
        return 75

def get_mile_rate(code):
    if code == "B":
        return 0.25
    elif code == "S":
        return 0.30
    elif code == "L":
        return 0.40

def calculate_miles(start_odemeter, end_odemeter):
    return end_odemeter - start_odemeter

def calculate_charged_miles(total_miles):

    if total_miles > 100:
        return total_miles - 100
    else:
        return 0

def calculate_daily_total_bill(days_rented, daily_rate, charged_miles, mile_rate):
    daily_charge = days_rented * daily_rate
    milage_charge = charged_miles * mile_rate
    return daily_charge + milage_charge

# PROGRAM LOGIC

code = input("Enter classification code, (B, S, L) ").upper()

while code not in ["B", "S", "L"]:
    print("Invalid classification code. Please enter B, S, or L.")
    code = input ("Enter classification code (B, S, L): ").upper()

start_odemeter = int(input("Start Odemeter: "))
end_odemeter = int(input("End Odemeter: "))
days_rented = int(input("Days Rented: "))

total_miles = calculate_miles(start_odemeter, end_odemeter)
charged_miles = calculate_charged_miles(total_miles)

classification_name = get_classification_name(code)
daily_rate = get_daily_rate(code)
mile_rate = get_mile_rate(code)

total_bill = (days_rented * daily_rate) + (charged_miles * mile_rate)

print(f"Classification: {classification_name}")
print("Classification:", classification_name)
print("total Miles:", total_miles)
print("Charged Miles:", charged_miles)
print(f"Total Bill: ${total_bill:.2f}")




```
