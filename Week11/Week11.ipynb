```python
# Name: rectangle.py
# AUTHOR: KeiShara
# IMPORT STATEMENTS

# FUNCTION DEFINITIONS

#  Function: calculate_rectangle(length, width)
#  Multiply length and width to find the area
#  Add length and width, then multiply by 2 to find the perimeter
#  Return both values together as a tuple (area,perimeter)

# PROGRAM LOGIC

# Ask the user to enter a length and width
# Convert the inputs into float values
# Call the calculate_rectangle function using those values
# Store the returned values into two variables
# Display the area and perimeter clearly

def calculate_rectangle(length, width):
    area = length * width
    perimeter= 2 * (length + width)
    return (area, perimeter)

length = float(input("Enter length: "))
width = float(input("Enter width:"))

my_area, my_perimeter = calculate_rectangle(length,width)

print("The Area is:", my_area)
print("The Perimeter is:", my_perimeter)
```
